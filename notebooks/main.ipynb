Building upon the Euclidean distance idea\
By creating matrices per player, which could be seen as a collection of vectors, lets say...
$$
M_1 = \begin{bmatrix} v_1  v_2  \cdots  v_n \end{bmatrix},\quad M_2 = \begin{bmatrix} u_1  u_2  \cdots  u_n \end{bmatrix},\quad \dots \quad M_i = \begin{bmatrix} w_1  w_2  \cdots  w_n \end{bmatrix}
$$
We could find the distance between EACH vector inside the matrices.\
We'd ultimately be left with a matrix showing distance...

Let's suppose D1 is just some random player. Each value d1, d2, is showing how far that specific overall statistic is from the chosen player. We would be able to compare these values among all the other ones computed, and the closest ones are those with the lowest d values.
$$
D_1 = \begin{bmatrix}d_1 d_2 \cdots d_n\end{bmatrix}
$$

Metrics like G/SoT, and Dist risk a NaN value not because the value is genuinely missing, but due to the absence of SoT, Gls, or Sh values. This causes a 0/0 division which returns NaN. We now move on to scaling the data. To continue with the original idea, how we'd have matrices...

To compare each 2000+ players would be highly inefficient, as we'd have to go through and perform the distance formula that many times (2000+). Rather, what we can do, is create scaled groups. That is, just as we made the feature groups for v1, we will keep the same groups, and they will be mapped to a DataFrame, containing the scaled values for those columns. Indices stay the same, only thing that changes is the scaled values.

From this, what we can do is set a target vector. That target vector will be the player that we are looking for. Let's say its Lamine Yamal. We know his index is given by:

```python
pdata[pdata['Player']=='Lamine Yamal'].index
```

idx = 2792. So given that, we would set the target vector as that. We'd grab it from the DataFrame as mentioned above, say with feature group shooting, and we'd have the target as Lamine's shooting vectors, and X as everyone else. We then perform the Euclidean distance function. We will be returned a 1D array where the index corresponds to the DataFrame. We can say distances[`0`] corresponds to pdata.iloc[`0`]'s distance to pdata.iloc[`target_idx`]

```python
scaled_groups = {}

for group_name, cols in feature_groups_v1.items():
    group_data = pdata_n[cols].copy()

    scaler = StandardScaler()

    group_scaled = pd.DataFrame(
        scaler.fit_transform(group_data),
        columns=cols,
        index=group_data.index
    )

    scaled_groups[group_name] = group_scaled
```

```python
def dist_to_target(X, target_idx):
    target_vector = X.iloc[target_idx].values
    distances = np.linalg.norm(X.values - target_vector, axis=1)
    return distances
```

```python
def perform_check(name, scaled_groups, feature_cat):
    matches = pdata[pdata["Player"] == name]
    if matches.empty:
        raise ValueError(f'No player found with name: {name}')
    target_idx = pdata[pdata['Player'] == name].index[0]
    X = scaled_groups[feature_cat]
    return dist_to_target(X, target_idx)
```

```python
name = input("Enter Player Name: ")
checked = {}
similar_indices = []
run = 0
for feature_cat in scaled_groups:
    run += 1
    distances = perform_check(name, scaled_groups, feature_cat)
    top5_idx = np.argsort(distances)[:5]
    top5_values = distances[top5_idx]
    print(f'Top 5 idx at run {run}, i.e. {feature_cat}: {top5_idx}')  
    similar_indices.append(top5_idx)
    #similar_indices = np.append(similar_indices, (top5_idx))
    
similar_indices = np.array(similar_indices)
similar_indices.shape
```
