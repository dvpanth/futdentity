Briefly looking at the dataframe, I notice there are many columns that seem redundant, due to being duplicates. Columns including Age, 90s, Gls, etc. are often duplicated bloating the columns to its 267 value, when it could be less. An idea to calculate player similarity is to utilise the Lin. Alg. concept of distance, i.e. between vectors. By making vectors, we can calculate Euclidean distance and therefore how far apart these players are statistically.

We'll gather all the quantitative and qualitative columns, sift through to identify duplicates

```python
''' Select numeric and categorical columns from dataset '''
num_cols = pdata.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = pdata.select_dtypes(include=['str']).columns.tolist()
```

#### Preprocessing and Data Cleaning

Removed redundant/duplicate columns

```python
''' Drop columns duplicate in nature '''
duplicate_tokens = ['Rk', 'Age', 'Born', '90s', 'Gls', 'PK', 'PKatt', 'xG', 'npxG', 'Ast', 'xAG', 'PrgP', 'Att', 'Crs',
                    'Off', 'Blocks', 'Fld', 'Sh', 'Int', 'TklW', 'PrgC', 'PrgR', 'TotDist', 'PrgDist', '1/3', 'MP',
                    'Min', 'Starts', 'CrdY', 'CrdR', 'Lost']
for token in duplicate_tokens:
    pdata = pdata.loc[:, pdata.columns.str.contains(f'^{token}$') | ~pdata.columns.str.contains(token)]

pdata = pdata.drop(columns=['Rk']) # Redundant Rk column, utilize native n=0 indexing instead

pdata.shape
```

Update num_columns to fit the changes of the duplicate

```python
num_cols =  pdata.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = pdata.select_dtypes(include=['str']).columns.tolist()
```

Building upon the Euclidean distance idea\
By creating matrices per player, which could be seen as a collection of vectors, lets say...
$$
M_1 = \begin{bmatrix} v_1  v_2  \cdots  v_n \end{bmatrix},\quad M_2 = \begin{bmatrix} u_1  u_2  \cdots  u_n \end{bmatrix},\quad \dots \quad M_i = \begin{bmatrix} w_1  w_2  \cdots  w_n \end{bmatrix}
$$
We could find the distance between EACH vector inside the matrices.\
We'd ultimately be left with a matrix showing distance...

Let's suppose D1 is just some random player. Each value d1, d2, is showing how far that specific overall statistic is from the chosen player. We would be able to compare these values among all the other ones computed, and the closest ones are those with the lowest d values.
$$
D_1 = \begin{bmatrix}d_1 d_2 \cdots d_n\end{bmatrix}
$$

We will proceed with creating the feature groups. From num_cols, we have identified the group that each feature variable will be going into.

***One thing to outline*** is the efficiency and ratio metrics. Metrics like G/SoT, and Dist risk a NaN value not because the value is genuinely missing, but due to the absence of SoT, Gls, or Sh values. This causes a 0/0 division which returns NaN. NaN will harm the Euclidean distance calculations, thus, V1 will tag the efficiency and ratio metrics, and holds them aside during the scaling process.

```python
feature_groups_full = {
    "general_playing_time": [
        "MP", "Starts", "Min", "90s"
    ],

    "shooting": [
        "Gls", "Ast", "G+A",
        "PK",
        "xG", "xAG",
        "Sh", "SoT", "SoT%", "SoT/90", "G/SoT",
        "Dist", "FK"
    ],

    "passing": [
        "Cmp", "Att", "Cmp%",
        "TotDist", "PrgDist",
        "PrgP",
        "xA", "KP", "1/3", "PPA",
        "Live", "Dead",
        "TB", "Sw", "Crs",
        "TI", "CK", "In", "Out", "Str",
        "Off", "Blocks"
    ],

    "chance_creation": [
        "SCA", "SCA90",
        "PassLive", "PassDead",
        "TO", "Fld", "Def",
        "GCA", "GCA90"
    ],

    "defense": [
        "Tkl", "TklW",
        "Def 3rd", "Mid 3rd",
        "Tkl%", "Lost",
        "Pass", "Int", "Clr", "Err"
    ],

    "possession_carrying": [
        "Touches",
        "Def Pen", "Def 3rd", "Mid 3rd",
        "Live",
        "Succ", "Succ%",
        "Tkld", "Tkld%",
        "Carries", "CPA",
        "Mis", "Dis", "Rec",
        "PrgC", "PrgR"
    ],

    "substitution_context": [
        "Mn/Start", "Compl",
        "Subs", "Mn/Sub", "unSub",
        "PPM",
        "onG", "onGA",
        "+/-", "+/-90"
    ],

    "fouls_recovery_aerials": [
        "Fls", "OG", "Recov",
        "Won", "Won%"
    ],

    "team_results": [
        "W", "D", "L"
    ],

    "goalkeeping": [
        "GA", "GA90",
        "SoTA", "Saves", "Save%",
        "CS", "CS%"
    ],

    "goalkeeping_advanced": [
        "GA_stats_keeper_adv",
        "FK_stats_keeper_adv",
        "CK_stats_keeper_adv",
        "OG_stats_keeper_adv",
        "/90",
        "Cmp_stats_keeper_adv",
        "Cmp%_stats_keeper_adv",
        "Thr",
        "Launch%",
        "AvgLen",
        "Opp",
        "Stp", "Stp%",
        "#OPA", "#OPA/90",
        "AvgDist"
    ]
}

feature_tags = {
    # Shooting ratios / conditional stats
    "SoT%": ["efficiency", "percentage", "conditional", "denominator:Sh"],
    "SoT/90": ["rate", "per90"],
    "G/SoT": ["efficiency", "ratio", "conditional", "denominator:SoT"],
    "Dist": ["conditional", "denominator:Sh"],

    # Passing efficiency
    "Cmp%": ["efficiency", "percentage", "denominator:Att"],

    # Chance creation rates
    "SCA90": ["rate", "per90"],
    "GCA90": ["rate", "per90"],

    # Defensive efficiency
    "Tkl%": ["efficiency", "percentage", "conditional"],

    # Possession / take-on efficiency
    "Succ%": ["efficiency", "percentage", "conditional"],
    "Tkld%": ["efficiency", "percentage", "conditional"],

    # Substitution / playing-time context rates
    "Mn/Start": ["rate", "context"],
    "Mn/Sub": ["rate", "context"],
    "PPM": ["rate", "team_context"],
    "+/-90": ["rate", "per90", "team_context"],

    # Aerial efficiency
    "Won%": ["efficiency", "percentage", "conditional"],

    # Goalkeeping efficiency / rates
    "GA90": ["rate", "per90", "goalkeeping"],
    "Save%": ["efficiency", "percentage", "conditional", "denominator:SoTA"],
    "CS%": ["efficiency", "percentage", "conditional", "denominator:Starts"],

    # Advanced GK
    "/90": ["rate", "per90", "goalkeeping"],
    "Cmp%_stats_keeper_adv": ["efficiency", "percentage", "goalkeeping"],
    "Launch%": ["percentage", "goalkeeping"],
    "Stp%": ["efficiency", "percentage", "conditional", "goalkeeping"],
    "#OPA/90": ["rate", "per90", "goalkeeping"],
}
```

```python
exclude_tags_v1 = {"efficiency", "percentage", "ratio", "conditional"} # Tags to IGNORE

feature_groups_v1 = {
    group: [
        col for col in cols 
        if not any(tag in exclude_tags_v1 for tag in feature_tags.get(col, []))
        ] 
    for group, cols in feature_groups_full.items()
}
```

Testing scaling the shooting data

```python
shooting_cols = feature_groups_v1["shooting"]
shooting_data = pdata_n[shooting_cols].copy()
shooting_scaler = StandardScaler()
shooting_standardized = pd.DataFrame(
    shooting_scaler.fit_transform(shooting_data),
    columns=shooting_cols,
    index=shooting_data.index,
)
shooting_standardized.head()
```

```python
scaled_groups = {}

for group_name, cols in feature_groups_v1.items():
    group_data = pdata_n[cols].copy()

    scaler = StandardScaler()

    group_scaled = pd.DataFrame(
        scaler.fit_transform(group_data),
        columns=cols,
        index=group_data.index
    )

    scaled_groups[group_name] = group_scaled
```
