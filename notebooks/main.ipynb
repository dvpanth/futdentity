#### Preprocessing and Data Cleaning

Removed redundant/duplicate columns

```python
''' Drop columns duplicate in nature '''
duplicate_tokens = ['Rk', 'Age', 'Born', '90s', 'Gls', 'PK', 'PKatt', 'xG', 'npxG', 'Ast', 'xAG', 'PrgP', 'Att', 'Crs',
                    'Off', 'Blocks', 'Fld', 'Sh', 'Int', 'TklW', 'PrgC', 'PrgR', 'TotDist', 'PrgDist', '1/3', 'MP',
                    'Min', 'Starts', 'CrdY', 'CrdR', 'Lost']
for token in duplicate_tokens:
    pdata = pdata.loc[:, pdata.columns.str.contains(f'^{token}$') | ~pdata.columns.str.contains(token)]

pdata = pdata.drop(columns=['Rk']) # Redundant Rk column, utilize native n=0 indexing instead

pdata.shape
```
