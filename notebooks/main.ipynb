Building upon the Euclidean distance idea\
By creating matrices per player, which could be seen as a collection of vectors, lets say...
$$
M_1 = \begin{bmatrix} v_1  v_2  \cdots  v_n \end{bmatrix},\quad M_2 = \begin{bmatrix} u_1  u_2  \cdots  u_n \end{bmatrix},\quad \dots \quad M_i = \begin{bmatrix} w_1  w_2  \cdots  w_n \end{bmatrix}
$$
We could find the distance between EACH vector inside the matrices.\
We'd ultimately be left with a matrix showing distance...

Let's suppose D1 is just some random player. Each value d1, d2, is showing how far that specific overall statistic is from the chosen player. We would be able to compare these values among all the other ones computed, and the closest ones are those with the lowest d values.
$$
D_1 = \begin{bmatrix}d_1 d_2 \cdots d_n\end{bmatrix}
$$

Metrics like G/SoT, and Dist risk a NaN value not because the value is genuinely missing, but due to the absence of SoT, Gls, or Sh values. This causes a 0/0 division which returns NaN. ### Data Scaling, Comparisons

We now move on to scaling the data. To continue with the original idea, how we'd have matrices...

To compare each 2000+ players would be highly inefficient, as we'd have to go through and perform the distance formula that many times (2000+). Rather, what we can do, is create scaled groups. That is, just as we made the feature groups for v1, we will keep the same groups, and they will be mapped to a DataFrame, containing the scaled values for those columns. Indices stay the same, only thing that changes is the scaled values.

From this, what we can do is set a target vector. That target vector will be the player that we are looking for. Let's say its Lamine Yamal. We know his index is given by:

```python
pdata[pdata['Player']=='Lamine Yamal'].index
```

idx = 2792. So given that, we would set the target vector as that. We'd grab it from the DataFrame as mentioned above, say with feature group shooting, and we'd have the target as Lamine's shooting vectors, and X as everyone else. We then perform the Euclidean distance function. We will be returned a 1D array where the index of each element corresponds to the index on the DataFrame. We can say distances[`0`] corresponds to pdata.iloc[`0`]'s distance to pdata.iloc[`target_idx`]

```python
scaled_groups = {}

for group_name, cols in feature_groups_v1.items():
    group_data = pdata_n[cols].copy()

    scaler = StandardScaler()

    group_scaled = pd.DataFrame(
        scaler.fit_transform(group_data),
        columns=cols,
        index=group_data.index
    )

    scaled_groups[group_name] = group_scaled
```

```python
def dist_to_target(X, target_idx):
    target_vector = X.iloc[target_idx].values
    distances = np.linalg.norm(X.values - target_vector, axis=1)
    return distances
```

```python
def get_target_idx(name, pdata):
    matches = pdata[pdata["Player"].str.casefold() == name.casefold()]

    if matches.empty:
        raise ValueError(f"No player found matching: {name}")

    if len(matches) > 1:
        detail_cols = [col for col in ["Squad", "Comp", "Pos"] if col in matches.columns]
        options = matches[["Player", *detail_cols]].reset_index(drop=True)

        print("Multiple players found:")

        for number, (_, player) in enumerate(options.iterrows(), start=1):
            details = ", ".join(f"{col}: {player[col]}" for col in detail_cols)
            print(f"{number}. {player['Player']} ({details})")

        while True:
            choice = input(f"Select the player number (1-{len(matches)}): ").strip()

            if choice.isdigit() and 1 <= int(choice) <= len(matches):
                return matches.index[int(choice) - 1]

            print(f"Enter a number from 1 to {len(matches)}.")

    return matches.index[0]
```

With the functions defined, we will test using again, Lamine Yamal, specifically, his possession_carrying statistic, to visualize the differences.

```python
target_idx = get_target_idx('Lamine Yamal', pdata)
X = scaled_groups['possession_carrying']

distances = dist_to_target(X, target_idx)
closest_idx = np.argsort(distances)
closest_idx = closest_idx[closest_idx != target_idx][:5]

comparison = pd.DataFrame(
    [X.iloc[target_idx], *[X.iloc[i] for i in closest_idx]],
    index=["Lamine Yamal", *[pdata["Player"].iloc[i] for i in closest_idx]],
)

plt.figure(figsize=(max(8, len(X.columns)), 2.5))
sns.heatmap(comparison, center=0, cmap="vlag", annot=True, fmt=".2f")
plt.title("Standardized possession and carrying features")
plt.xlabel("Feature")
plt.ylabel("Player")
plt.tight_layout()
plt.show()
```

Visually, we can see te differences in players. Let's look specifically at Kylian Mbappe. We notice his individual features are in fact, not too far from Lamine's. Alot of them are within ~1 standard deviation away. That is, except for Successful Dribbles, where Lamine is a beast. This, accumulated with the differences in the other stats, create that difference. Mbappe is the closest one to Lamine.

This is the same check that is done for shooting as well. We can check the shooting one below. It will not be the same as possession, as players are multifaceted, thus, they will have different strengths, weaknesses, resulting in different stats:

```python
target_idx = get_target_idx('Lamine Yamal', pdata)
X = scaled_groups['shooting']

distances = dist_to_target(X, target_idx)
closest_idx = np.argsort(distances)
closest_idx = closest_idx[closest_idx != target_idx][:5]

comparison = pd.DataFrame(
    [X.iloc[target_idx], *[X.iloc[i] for i in closest_idx]],
    index=["Lamine Yamal", *[pdata["Player"].iloc[i] for i in closest_idx]],
)

plt.figure(figsize=(max(8, len(X.columns)), 2.5))
sns.heatmap(comparison, center=0, cmap="vlag", annot=True, fmt=".2f")
plt.title("Standardized shooting features")
plt.xlabel("Feature")
plt.ylabel("Player")
plt.tight_layout()
plt.show()
```
